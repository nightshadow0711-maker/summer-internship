```python
mod1['Tweet'] = mod1['Tweet'].astype(str)
rahul['Tweet'] = mod1['Tweet'].astype(str)

def find_polarity(review):
  return TextBlob(review).sentiment.polarity
```

```python
mod1['Polarity'] = mod1['Tweet'].apply(find_polarity)
rahul['Polarity'] = rahul['Tweet'].apply(find_polarity)
```

```python
def get_sentiment_label(polarity):
    if polarity > 0:
        return 'Positive'
    elif polarity < 0:
        return 'Negative'
    else:
        return 'Neutral'

rahul['Label'] = rahul['Polarity'].apply(get_sentiment_label)
rahul_count = rahul.groupby('Label').count()
```

```python
# Ensure 'Label' column exists for mod1 and calculate sentiment counts
mod1['Label'] = mod1['Polarity'].apply(get_sentiment_label)
mod1_count_by_label = mod1.groupby('Label').count()

# Get sentiment counts for Modi
neg_modi = mod1_count_by_label.loc['Negative', 'Tweet']
pos_modi = mod1_count_by_label.loc['Positive', 'Tweet']

# Get sentiment counts for Rahul (rahul_count is already by Label as per kernel state)
neg_rahul = rahul_count.loc['Negative', 'Tweet']
pos_rahul = rahul_count.loc['Positive', 'Tweet']

# Correct politician list syntax
politicians = ["Modi", "Rahul"]

# Create lists for plotting with correct variables
neg_list = [neg_modi, neg_rahul]
pos_list = [pos_modi, pos_rahul]

# Create the bar chart with corrected typos
fig = go.Figure(data=[
    go.Bar(name="Negative", x=politicians, y=neg_list),
    go.Bar(name="Positive", x=politicians, y=pos_list)
])
fig.show()
```
